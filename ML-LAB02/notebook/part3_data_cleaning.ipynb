```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv("../data/Loan_Default.csv")
df.head()
```

```python
# เช็คก่อนว่ามี missing ตรงไหน
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print(f"Missing values ที่เหลือ: {df.isnull().sum().sum()}")
```

```python
print(f"ก่อนลบ: {df.shape[0]} แถว")
print(f"จำนวนที่ซ้ำ: {df.duplicated().sum()}")

df = df.drop_duplicates()

print(f"หลังลบ: {df.shape[0]} แถว")
```

```python

# ถ้า year ควรเป็น int แต่ดันเป็น float
df['year'] = df['year'].astype(int)

# ถ้า Status ควรเป็น category แทน int
df['Status'] = df['Status'].astype('category')
print(df.dtypes)
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
print(numeric_cols.tolist())
```

```python
comparison = pd.DataFrame({
    'Mean': df[numeric_cols].mean(),
    'Median': df[numeric_cols].median(),
    'Difference': df[numeric_cols].mean() - df[numeric_cols].median()
})
print(comparison)
```
